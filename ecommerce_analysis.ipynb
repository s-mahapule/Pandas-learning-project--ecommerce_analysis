Phase 2

```python
#Determine the total revenue generated by the business, then identify the products that generate the most revenue.
```

```python
#total revenue generated by the business
data["Revenue"] = data["Price"] * data["Quantity"]
revenue_total = data["Revenue"].sum()
print(f"Total Revenue :{revenue_total:.2f}")
```

```python
# identify the products that generate the most revenue.
data.groupby(["StockCode", "Description"]).Revenue.sum().sort_values(ascending = False)
```

```python
#How much of the revenue calculation comes from cancelled transactions, and how does total revenue change if cancellations are excluded?
```

```python
cancelled_tra_total = data.loc[data["Invoice"].astype(str).str.startswith("C"), "Revenue"].sum()
print(f"Total Revenue of cancelled transactions :{cancelled_tra_total:.2f}")
```

```python
print(F"Revenue after excluding cancellations:  { revenue_total - abs(cancelled_tra_total):.2F}")
```
